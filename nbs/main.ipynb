```python
import os, glob
from pathlib import Path
```

```python
model_size = 'small'
files = glob.glob(f'../outputs/lm1b-subset=*-{model_size}')
files.sort(key=lambda f: float(Path(f).name.split('=')[1].split('-')[0]))
files
```

```python
c = 0
store_to = True
for f in files:
    ckpts = glob.glob(os.path.join(f, 'checkpoints/step_*'))
    try:
        ckpts.sort(key = lambda f: int(Path(f).name.split('_')[1].split('.')[0]))
    except:
        print(f'Fail: {f}')
        
    if len(ckpts) == 0: continue
        
    max_file = Path(ckpts[-1]).name.split('.')[0].split('_')[1]
    if int(max_file) == 1_000_000:
        subset = float(Path(f).name.split('=')[1].split('-')[0])
        
        if not os.path.exists(f'../model-ckpts/lm1b-{model_size}/lm1b-{model_size}-{subset}.ckpt'):
            os.system(f'cp {ckpts[-1]} ../model-ckpts/lm1b-{model_size}/lm1b-{model_size}-{subset}.ckpt')
    else:
        c += 1
        print(f, max_file)
        
print(f'Remaining Num. of Trained Models: {c}') 
```

```python
def filter_ckpts(ckpts):
    new_ckpts = []
    keep_idx = set(
        [2500] + [i for i in range(0, 1_050_000, 25_000)]
    )
    for f in ckpts:
        step = int(Path(f).name.split('_')[1].split('.')[0])
        if step in keep_idx:
            continue
        new_ckpts.append(f)
    return new_ckpts
```

```python
for f in files:
    ckpts = glob.glob(os.path.join(f, 'checkpoints/step_*'))

    if len(ckpts) <= 30:
        print(f'Skipping... {f}')
        continue

    ckpts.sort(key = lambda f: int(Path(f).name.split('_')[1].split('.')[0]))
    ckpts = filter_ckpts(ckpts)
    
    to_del_size = int(len(ckpts) * 0.95)
    print(f'Deleting {to_del_size}/{len(ckpts)} files.')
    for j in range(to_del_size):
        to_del_f = ckpts[j]
        os.system(f'rm {to_del_f}')
```
